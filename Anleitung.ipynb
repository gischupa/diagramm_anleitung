Möglich wäre z.B. ein Gitternetz und ein automatisches Speichern des Bildes als Grafik:

```

plt.plot(t, s)

plt.show()
```

## Vorteile gegenüber Excel (oder allgemein Tabellenkalkulationen)
* Die Befehlsabfolge ist kurz und kopierbar. Du musst nur die Zahlenwerte austauschen.
* Kein Verklicken in Assistenten
* Die KI kann solchen Code erstellen
* Schnelles Erweitern des Diagramms:

```python
plt.grid(True)
plt.plot(t, s)
plt.scatter(t, s)
plt.show()
```

````python
# Programmfenster
````

Dass die Gitternlinien nicht durch die Punkte verlaufen, liegt an der automatischen Aufteilung der x-Achse.
Generell gelten in den Naturwissenschaften etwas andere Regeln für Diagramme, als bei den Graphen in Mathematik
